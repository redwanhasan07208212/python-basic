```python
temp = gap.pivot(index='country',columns='year',values='lifeExp')
temp
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(temp)
```

```python
temp2 = gap[gap['continent']=='Europe']
temp2
```

```python
new_temp = temp2.pivot(index='country',columns='year',values='lifeExp')
new_temp
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(new_temp,annot=True)
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(new_temp,annot=True,linewidths=0.5)
```

```python
new_temp = gap[gap['continent']=='Europe'].pivot(index='country',columns = 'year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(new_temp,annot=True,linewidths=0.5,cmap='summer')
```
