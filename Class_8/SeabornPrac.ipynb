```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd
```

```python
tips = pd.read_csv('tips.csv')
tips
```

```python
sns.scatterplot(x=tips['tip'],y=tips['total_bill'])
```

```python
sns.scatterplot(tips,x='tip',y='total_bill')
```

```python
sns.scatterplot(data=tips,x='tip',y='total_bill')
```

```python
sns.relplot(tips,x='tip',y='total_bill')
```

```python
sns.relplot(tips,x='tip',y='total_bill',kind='line')
```

```python
sns.relplot(data=tips,x='tip',y='total_bill',hue='sex')
```

```python
sns.relplot(tips,x='tip',y='total_bill',kind='scatter',hue='sex')
```

```python
sns.relplot(tips,x='tip',y='total_bill',kind='scatter',hue='sex',style='time')
```

```python
sns.relplot(tips,x='tip',y='total_bill',kind='scatter',hue='sex',style='time',size='size')
```
