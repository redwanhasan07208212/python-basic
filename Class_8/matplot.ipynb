```python
import pandas as pd 
import numpy as np
```

```python
data1 = pd.read_excel('linechart.xlsx',sheet_name='product sales')
data1.head()
```

```python
plt.figure(figsize=(16,6))
plt.plot(data1['Quarter'],data1['Dishwasher'],label = 'DishWasher Sales',color='red')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(16,6))
plt.plot(data1['Quarter'],data1['Dishwasher'], label = 'Dishwasher Sales' , color = "red" , marker = 'o')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(16,6))
plt.plot(data1['Quarter'],data1['Dishwasher'],label = 'Dishwasher Sales', color = 'red', marker='o', markersize = 8,linewidth = 2 , markerfacecolor = 'black' , markeredgecolor = 'black' , linestyle = '--' )
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(16,6))
plt.plot(data1['Quarter'],data1['Dishwasher'],label = 'Dishwasher Sales',color='red',marker='o',markersize=8,linewidth=2,markerfacecolor = 'black')
plt.legend()
plt.title('Dishwasher Sales over Quarter')
plt.xlabel('Quarter',fontsize = 16)
plt.ylabel('Sales Amount',fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(16,6))
plt.plot(
    data1['Quarter'],
    data1['Dishwasher'],
    label='Dishwasher Sales',
    color='red',
    marker='o',
    markersize=8,
    linewidth=2,
    markerfacecolor='black'
)


# Add grid
plt.grid(True, linestyle='--', alpha=0.5)

# Set background color
plt.gca().set_facecolor('#f9f9f9')  # light gray background
plt.gcf().patch.set_facecolor('white')  # figure background

# Title and labels
plt.title('📈 Dishwasher Sales Over Quarters', fontsize=20, fontweight='bold')
plt.xlabel('Quarter', fontsize=14)
plt.ylabel('Sales Amount', fontsize=14)

# Customize ticks
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# Show legend
plt.legend(fontsize=12)


plt.show()
```

```python
plt.figure(figsize=(16, 6))

# Plotting with simple enhancements
plt.plot(data1["Quarter"],
         data1["Dishwasher"],
         label="Dishwasher Sales",
         color="red",
         linewidth=2.5,
         marker="o",
         markersize=8,
         markerfacecolor="white",
         markeredgecolor="black",
         )


# Setting a black background color
plt.gca().set_facecolor('black')
plt.gcf().set_facecolor('black')

# Titles and labels
plt.title('Dishwasher Sales Over Quarters', fontsize=18,color="white")
plt.xlabel('Quarter', fontsize=14,color="white")
plt.ylabel('Sales Amount', fontsize=14,color="white")

# Show grid for readability
plt.grid(True, linestyle=':', alpha=0.4)

# Add legend
plt.legend(fontsize=12)


# Show the plot
plt.tight_layout()
plt.show()
```
