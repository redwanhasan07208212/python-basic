```python
df = pd.read_csv('batter.csv')
df = df.head(50)
df
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['avg'],df['strike_rate'],color='red')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.title('Avg and SR analysis between top 50 batsman')
```

```python
plt.figure(figsize=(12, 8))

# Use Seaborn style for better visuals
sns.set(style="whitegrid")

# Scatter with size and color
scatter = plt.scatter(
    df['avg'],
    df['strike_rate'],
    c=df['strike_rate'],             # Color by strike rate
    s=df['avg'] * 2,                 # Size by average
    cmap='viridis',                 # Color map
    alpha=0.7,
    edgecolors='k'
)

# Add colorbar
plt.colorbar(scatter, label='Strike Rate')

# Add grid and titles
plt.xlabel('Batting Average', fontsize=14)
plt.ylabel('Strike Rate', fontsize=14)
plt.title('Batting Average vs Strike Rate of Top 50 Batsmen', fontsize=16)
plt.grid(True, linestyle='--', alpha=0.5)

# Optional: Highlight top 3 strike rates
top_sr = df.nlargest(3, 'strike_rate')
for i, row in top_sr.iterrows():
    plt.text(row['avg']+0.3, row['strike_rate']+0.3, row['batter'], fontsize=10, color='black')

plt.tight_layout()
plt.show()
```
